```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

data = pd.read_csv(r"C:\Users\aleks\Downloads\wine_cleared (1)\wine_cleared.csv")
data.head()
```

```python
data['price_round'] = data['price'].round().astype(int)
```

```python
regex = '\d{4}' # регулярное выражение для нахождения чисел
data['year'] = data['title'].str.findall(regex).str.get(0)
```

```python
data['is_usa'] = data['country'].apply(lambda x: 1 if x == 'US' else 0)
```

```python
data[data["country"] != "US"]["country"].value_counts()
```

```python
data['is_france'] = (data['country'] == 'France').astype(int)
data['is_italy']  = (data['country'] == 'Italy').astype(int)

#data['is_france'].sum()
data['is_italy'].sum()
```

```python
# преобразуем признак year в datetime
data['year'] = pd.to_datetime(data['year'], errors='coerce')

# создаём бинарный признак old_wine: 1, если год < 2010 и значение не пустое
data['old_wine'] = data['year'].apply(
    lambda x: 1 if (pd.notnull(x) and x.year < 2010) else 0
)

data['old_wine'].sum()
```

```python
data['locality'] = data['title'].str.extract(r'\((.*?)\)')
```
