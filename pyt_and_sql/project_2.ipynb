```python
#5.2
query = f"""
SELECT a.name,
       COUNT(e.id) AS employers_cnt
FROM areas a
LEFT JOIN vacancies v ON v.area_id = a.id
LEFT JOIN employers e ON e.area = a.id
WHERE v.id IS NULL
GROUP BY a.id, a.name
ORDER BY employers_cnt DESC
LIMIT 1;
"""
df = pd.read_sql_query(query, connection)
df
```

```python
#5.7
query = f"""
WITH y AS (
    SELECT a.name AS area, COUNT(*) AS cnt
    FROM vacancies v
    JOIN employers e ON v.employer_id = e.id
    JOIN areas a ON v.area_id = a.id
    WHERE e.name = 'Яндекс'
      AND a.name IN (
          'Москва', 'Санкт-Петербург', 'Новосибирск', 'Екатеринбург',
          'Казань', 'Красноярск', 'Нижний Новгород', 'Челябинск',
          'Уфа', 'Краснодар', 'Самара', 'Ростов-на-Дону',
          'Омск', 'Воронеж', 'Пермь', 'Волгоград'
      )
    GROUP BY a.name
)
SELECT *
FROM y

UNION ALL

SELECT 'Total' AS area, SUM(cnt) AS cnt
FROM y;
"""
df = pd.read_sql_query(query, connection)
df
```

```python
#6.6
query = """
SELECT experience,
       ROUND(AVG(
           (COALESCE(salary_from, salary_to) + COALESCE(salary_to, salary_from)) / 2.0
       )) AS avg_salary
FROM vacancies
WHERE (
    name ILIKE '%data scientist%'
    OR name ILIKE '%data science%'
    OR name ILIKE '%исследователь данных%'
    OR (name LIKE '%ML%' AND name NOT ILIKE '%HTML%')
    OR name ILIKE '%machine learning%'
    OR name ILIKE '%машинн%обучен%'
)
AND (salary_from IS NOT NULL OR salary_to IS NOT NULL)
GROUP BY experience
ORDER BY experience;
"""
df = pd.read_sql_query(query, connection)
df
```
